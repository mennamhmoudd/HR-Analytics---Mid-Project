### Feature Engineering

```python
df['2025 Salary'] = df['Gross Salary']-(df['January 2026 Raise']+df['Additional Raise During 2026'])
df['January 2026 Raise Rate'] = round((df['January 2026 Raise']+df['Additional Raise During 2026']) / df['2025 Salary'],2)
df
```

```python
df['Absence Deduction Amount']=(df['Absence Penalty (Days)']+df['Absence Days'])*df['Daily Wage']
df['Statutory Penalty Amount']=pd.to_numeric(df['Statutory Penalty (Days)'])*df['Daily Wage']
df['Administrative Penalty Amount']=df['Administrative Penalty (Days)']*df['Daily Wage']
df['Total Penalties'] = df['Absence Deduction Amount'] + df['Statutory Penalty Amount']+df['Administrative Penalty Amount']+df['Other Monetary Penalties']
df['Net Salary'] = df['Total Salary']-df['Total Penalties']

df
```

```python
df['Salaries Month'] = df['Month'].dt.month_name()
df['Salaries Month'] 
```

```python
df['Month_num'] = df['Month'].dt.month
df['Month_num']
```

##### Cumulative Increasement 

```python
# Calculate the increasement for each employee

df=df.sort_values(['Employee ID' , 'Month_num'])
df['Monthly Increase'] = (
    df.groupby(['Employee ID'])['Additional Raise During 2026'].diff()
    .fillna(df['Additional Raise During 2026'])
)

df
```

```python
total_increase_per_month = (
    df.groupby(['Month_num', 'Salaries Month'])['Monthly Increase'].sum()
    .reset_index()
    .sort_values('Month_num')
)

total_increase_per_month
```

```python
px.bar(data_frame=total_increase_per_month,
    x='Salaries Month',
    y='Monthly Increase',
    title = 'Raises per Months',
    labels = {

        'Salaries Month' : 'Months',
        'Monthly Increase': 'Total Increasement'
   
         },
    text_auto= True)
```
